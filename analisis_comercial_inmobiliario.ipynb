# 📊 Análisis Comercial Inmobiliario

El sector inmobiliario necesita evaluar su **desempeño comercial** para comprender su crecimiento, rentabilidad y comportamiento de clientes.

Se solicita construir un **dashboard ejecutivo** que permita analizar ventas, clientes y propiedades para apoyar decisiones estratégicas basadas en datos.

# 🎯 Preguntas que el dashboard responde:

**Desempeño general**

- ¿Cuál es el **ingreso total** generado por la empresa?
- ¿Cuántas **propiedades se han vendido**?
- ¿Cuál es el **precio promedio de venta**?
- ¿Cuál es la **comisión total** generada por la empresa?

**Análisis comercial**

- ¿Qué **tipo de propiedad genera más ingresos**?
- ¿Qué **segmento de clientes compra más propiedades**?
- ¿Qué canal de venta **genera mayor ingresos**?


**Análisis temporal**

- ¿Cómo evolucionan las **ventas en el tiempo**?
- ¿El negocio está **creciendo año contra año**?
- ¿Cómo va el desempeño **acumulado del año actual (YTD)**?

**Cohortes de clientes**

- ¿Los clientes **vuelven a comprar** después de su primera compra?
- ¿Qué **cohortes de clientes generan más compras/ingresos con el tiempo**?

# 📝 Resumen Ejecutivo

**Hallazgos clave:**

- El ingreso total del periodo fue de *$6,012 millones* generado por *8,500 ventas*.
  
- El tipo de propiedad que genera mayor revenue es "Casa" con *37.26%*.

- La ciudad con mayor volumen de ventas es "Ciudad de México" con *$3.24 mil M*.

- El canal de venta más eficiente en términos de ingresos es "Corredor" con *72.85%*.

---
**Métricas principales:**

- Ingreso Total: **$6,012,502,170**
  
- Cantidad de Ventas: **8,500**

- Ticket Promedio: **$707,353**

- Comisión Total: **$201 millones**

---
**Insights accionables:**

- El segmento de clientes "Primera vez" concentra la mayor parte del ingreso.


- Los clientes adquiridos en las cohortes de principios de 2023 (especialmente enero y febrero) muestran mayor recurrencia sostenida a lo largo de los meses siguientes.


- Las ventas muestran un crecimiento generalizado, cerrando 2024 con un crecimiento positivo de *111.14%*.

---
**Recomendaciones estratégicas**

- Priorizar la comercialización de propiedades tipo "Casa" y "Comercial", ya que juntas representan más del *69%* de los ingresos totales y tienen los tickets promedio más altos.


- Fortalecer el canal de ventas mediante "Corredor", que presenta una mayor participación en el revenue (casi 3/4 partes del total), posiblemente optimizando su esquema de comisiones.


- Diseñar programas de fidelización dirigidos específicamente al segmento "Primera vez" para capitalizar su alto volumen inicial y fomentar la recompra observada en las cohortes iniciales.
